# Water excretion event detection from body area changes

This code detects sudden decreases in Hydra body area from area time-series data.

Input:
An Excel file containing frame numbers and body area values. The area values may come from DeepLabCut body-outline coordinates analyzed by convex hull, or from an already calculated area file.

What the code does:

1. Uploads an Excel file in Google Colab.
2. Reads frame number and body area values for multiple recordings.
3. Calculates the frame-to-frame change in area.
4. Detects frames where the area suddenly decreases below a defined threshold.
5. Prints the candidate sudden-drop frames for each recording.

Output:
A printed list of frames where sudden body-area drops were detected.

Notes:
Sudden decreases in body area were used as candidate water-excretion events. The threshold should be adjusted depending on image scale, frame rate, tracking quality, and the size of the animal.

```python
import pandas as pd
from google.colab import files

# Manually upload the file
uploaded = files.upload()

# Assuming the uploaded file is an Excel file
file_name = list(uploaded.keys())[0]
data = pd.read_excel(file_name)

# Rename columns for clarity
data.columns = ['Frame', 'Data_0202', 'Data_05e', 'Data_04g_i']

# Calculate the difference between consecutive frames for each dataset
data['Diff_0202'] = data['Data_0202'].diff()
data['Diff_05e'] = data['Data_05e'].diff()
data['Diff_04g_i'] = data['Data_04g_i'].diff()

# Set a threshold to detect significant drops
threshold = -100  # Adjust this threshold based on your specific needs

# Identify frames where the drop exceeds the threshold
sudden_drops = {
    'Data_0202': data[data['Diff_0202'] < threshold],
    'Data_05e': data[data['Diff_05e'] < threshold],
    'Data_04g_i': data[data['Diff_04g_i'] < threshold],
}

# Display the frames with sudden drops
for key, drops in sudden_drops.items():
    diff_col_name = f'Diff_{key.split('_')[1]}" if key == 'Data_04g_i' else f'Diff_{key.split('_')[-1]}'
    print(f"Sudden drops in {key}:")
    print(drops[['Frame', key, diff_col_name]])
    print("\n")
```

# Random time-point generator

This code generates a random list of non-repeated numbers within a user-defined range. This is used for analysis of periodcity in Figure 1.

Input:
The user enters:

1. The end of the number range
2. The number of random values to generate

What the code does:

1. Creates possible values from 0 to the chosen end point, in 0.5-unit steps.
2. Randomly selects the requested number of values without replacement.
3. Sorts the selected values from smallest to largest.
4. Saves the values as a `.csv` file.
5. Downloads the file from Google Colab.

Output:
A file named:

`random_numbersRandom.csv`

Notes:
This was used to generate randomized time points or frame/time values for analysis. Because values are selected without replacement, the same value will not appear twice. The maximum number of values depends on the selected range and the 0.5-unit interval.


```python
import numpy as np
import pandas as pd
from google.colab import files  # Import files from google.colab

# Function to generate random numbers
def generate_random_numbers():
    end_of_range = float(input("Enter the end of the number range: "))
    num_values = int(input("How many values do you want to generate? "))

    possible_values = np.arange(0, end_of_range + 0.5, 0.5)

    if num_values > len(possible_values):
        raise ValueError("Number of values requested exceeds the number of unique possibilities.")

    random_values = np.random.choice(possible_values, size=num_values, replace=False)
    random_values_sorted = np.sort(random_values)

    # Define the original file name
    original_file_name = 'random_numbers'
    # Adding "Random" to the file name
    modified_file_name = original_file_name + "Random.csv"
    # Save the numbers to a CSV file
    pd.DataFrame(random_values_sorted).to_csv(modified_file_name, index=False, header=False)
    print(f"Random numbers saved to {modified_file_name}")

    # Trigger download
    files.download(modified_file_name)  # Automatically prompt for a file download

# Call the function to generate and download random numbers
generate_random_numbers()
```
